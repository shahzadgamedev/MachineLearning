## Logistic Regression Functions

### Sigmoid Function
The sigmoid function is used to map the output of the linear function to a probability value between 0 and 1.
The sigmoid function is defined as:
$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

### Loss Function
The loss function is used to calculate the error of only one sample.
The loss function is defined as:
$$L(y, \hat{y}) = -y \log(\hat{y}) - (1 - y) \log(1 - \hat{y})$$

###  Cost Function
The cost function is used to measure the error of the entire dataset.
The cost function is defined as:
$$J(w, b) = \frac{1}{m} \sum_{i=1}^{m} L(y^{(i)}, \hat{y}^{(i)})$$

Where:
- $m$ is the number of samples
- $y^{(i)}$ is the actual label of the $i^{th}$ sample
- $\hat{y}^{(i)}$ is the predicted label of the $i^{th}$ sample
- $w$ is the weight vector
- $b$ is the bias
- $L(y, \hat{y})$ is the loss function
- $J(w, b)$ is the cost function

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def model(x, w, b):
    return sigmoid(np.dot(x, w) + b)

def loss_function(y_hat, y):
    return (-y * np.log(y_hat) - (1 - y) * np.log(1 - y_hat)) / 2
    
def cost_function(x,y,w,b):
    m = x.shape[0]
    cost = 0.0
    for i in range(m):
        y_hat = model(x[i], w, b)
        cost += -y[i]*np.log(y_hat) - (1-y[i])*np.log(1-y_hat)
        
    cost /= m
    return cost
```

## Compute the Gradient
The gradient is used to update the weights and bias.
The gradient is defined as:
$$\frac{\partial J(w, b)}{\partial w} = \frac{1}{m} \sum_{i=1}^{m} (y^{(i)} - \hat{y}^{(i)}) x^{(i)}$$
$$\frac{\partial J(w, b)}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} (y^{(i)} - \hat{y}^{(i)})$$
    
    Where:
- $m$ is the number of samples
- $y^{(i)}$ is the actual label of the $i^{th}$ sample
- $\hat{y}^{(i)}$ is the predicted label of the $i^{th}$ sample
- $x^{(i)}$ is the $i^{th}$ sample
- $w$ is the weight vector
- $b$ is the bias
- $J(w, b)$ is the cost function
- $\frac{\partial J(w, b)}{\partial w}$ is the gradient of the cost function with respect to the weight vector
- $\frac{\partial J(w, b)}{\partial b}$ is the gradient of the cost function with respect to the bias

```python
def gradient(x, y, w, b):
    m,n = x.shape
    dj_dw = np.zeros((n,))
    dj_db = 0.0
    
    for i in range(m):
        y_hat = model(x[i], w, b)
        dj_dw += (y_hat - y[i]) * x[i]
        dj_db += y_hat - y[i]
    dj_dw /= m
    dj_db /= m
    return dj_dw, dj_db
```

## Gradient Descent
The gradient descent algorithm is used to update the weights and bias.
### Parameters
- $x$ is the feature matrix
- $y$ is the label vector
- $w$ is the weight vector
- $b$ is the bias
- $learning\_rate$ is the learning rate
- $num\_iterations$ is the number of iterations

### Steps
1. Calculate the gradient
2. Update the weights and bias
3. Repeat steps 1 and 2 until the cost function converges
4. Return the weights, bias, cost history, weight history, and bias history



```python
def gradient_descent(x, y, w, b, alpha, iterations):
    j_hist = []
    w_hist = np.zeros((iterations, w.shape[0]))
    b_hist = np.zeros((iterations, 1))
    
    for i in range(iterations):
        dj_dw, dj_db = gradient(x, y, w, b)
        
        
        for j in range(w.shape[0]):
            w[j] -= alpha * dj_dw[j]
        b -= alpha * dj_db
        
        if i < 100000:
            j_hist.append(cost_function(x, y, w, b))
        w_hist[i] = w
        b_hist[i] = b
    return w, b, j_hist, w_hist, b_hist
```
