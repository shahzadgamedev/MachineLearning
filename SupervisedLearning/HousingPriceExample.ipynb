# Regression Example
## Example of Linear Regression with Multiple Features
The purpose of this example is to understand the linear regression with multiple features.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the conversion functions
def convert_yes_no(value):
    return 1.0 if value.decode('utf-8') == 'yes' else 0.0

def convert_furnishing_status(value):
    value = value.decode('utf-8')
    if value == 'unfurnished':
        return 0.0
    elif value == 'semi-furnished':
        return 1.0
    elif value == 'furnished':
        return 2.0
    else:
        raise ValueError(f"Unknown furnishing status: {value}")

converters = {
    5: convert_yes_no,  # main road
    6: convert_yes_no,  # guestroom
    7: convert_yes_no,  # basement
    8: convert_yes_no,  # hot water heating
    9: convert_yes_no,  # air conditioning
    11: convert_yes_no, # pref area
    12: convert_furnishing_status  # furnishing status
}
# Load the data from the CSV file, skipping the header
data = np.genfromtxt(r'.\Regression\Data\Housing_Data.csv', delimiter=',', skip_header=1, dtype=float, converters=converters)

# Split the data into features (x) and target (y)
x = data[:, 1:]  # all columns except the first one
y = data[:, 0]   # first column

# plot x-axis as area and y-axis as price
plt.scatter(x[:, 0], y)
plt.xlabel('Area')
plt.ylabel('Price')
plt.ticklabel_format(style='plain', axis='y')
plt.show()


```

### Vectorized Cost Function

```python
def model(features, w, b):
    return np.dot(features, w) + b


def cost_function(features, target, params_w, param_b):
    m = features.shape[0]
    cost = np.sum((model(features, params_w, param_b) - target) ** 2)
    return (1 / (2 * m)) * cost

def scale_features(features):
    return features / 100
def scale_target(target):
    return target / 1000
def back_scale_features(features):
    return features * 100
def back_scale_target(target):
    return target * 1000

y = scale_target(y)

# only scale the area column
x[:, 0] = scale_features(x[:, 0])
```

### Gradient Descent with Multiple Features

```python
iterations = 10000
learning_rate = 0.0001
weight = np.zeros(x.shape[1])
bias = 0

cost_history = []
weight_history = np.zeros((iterations, x.shape[1]))
bias_history = np.zeros(iterations)


def calculate_gradient(features, target, params_w, param_b):
    m = features.shape[0]  # no of rows
    n = features.shape[1]  # no of Column

    d_w = np.zeros(n)
    d_b = 0
    for p in range(m):
        d_w += (model(features[p], params_w, param_b) - target[p]) * features[p]
        d_b += model(features[p], params_w, param_b) - target[p]

    # simultaneous update
    # Calculate average gradient
    d_w /= m
    d_b /= m
    return d_w, d_b


for i in range(iterations):
    columns = x.shape[1]
    temp_weight = weight
    temp_bias = bias
    
    # calculate the gradient before updating the weight
    derivative_w, derivative_b = calculate_gradient(x, y, weight, bias)
    
    #gradient descent for each feature(Column) of x
    for c in range(columns):
        temp_weight[c] = temp_weight[c] - learning_rate * derivative_w[c]
    temp_bias = bias - learning_rate * derivative_b
    
    #simultaneous update
    weight = temp_weight
    bias = temp_bias
    weight_history[i] = weight
    bias_history[i] = bias
    
    if i < 100000:
        cost_history.append(cost_function(x, y, weight, bias))

# plot the iterations vs cost
plt.subplots(1, 3, figsize=(30, 10))

plt.subplot(1,3,1)
plt.plot(range(iterations), cost_history)
plt.xlabel('Iterations')
plt.ylabel('Cost')


feature_names = [
    'area', 'bedrooms', 'bathrooms', 'stories',
    'main-road', 'guestroom', 'basement', 'hot-water-heating',
    'air-conditioning', 'parking', 'pref-area', 'furnishing-status'
]

# plot the iterations vs weight
plt.subplot(1, 3, 2)
for idx in range(len(feature_names)):
    plt.plot(range(iterations), weight_history[:, idx], label=feature_names[idx])
plt.xlabel('Iterations')
plt.ylabel('Weight')
plt.legend(loc='upper right')  # Add legend to distinguish each weight


# plot the iterations vs bias
plt.subplot(1,3,3)
plt.plot(range(iterations), bias_history)
plt.xlabel('Iterations')
plt.ylabel('Bias')

plt.tight_layout()
plt.show()

print(f'Weight: {repr(weight)}')
print(f'Bias: {bias}')
```

## Testing the Model
Using the trained model to predict the price of a house with an area of 2000 sqft, 4 bedrooms, 3 bathrooms, main road, guest room, basement, 
hot water heating, air conditioning, parking, pref area, and semi-furnished.

```python

test_data = {
    'area': 7420,
    'bedrooms': 4,
    'bathrooms': 2,
    'stories': 3,
    
    # binary features(yes/no) 0 = no, 1 = yes
    'main-road': 1,
    'guestroom': 0,
    'basement': 0,
    'hot-water-heating': 0,
    'air-conditioning': 1,
    'parking': 2,
    'pref-area': 1,
    'furnishing-status': 1 # 0: unfurnished, 1: semi-furnished, 2: furnished
}

# convert the test data to numpy array
test_data = np.array(list(test_data.values()))

# predict the price
predicted_price = model(test_data, weight, bias)

# back scale the price
predicted_price = back_scale_target(predicted_price)
print(f'The predicted price of the house is ${predicted_price:.2f}')
```
