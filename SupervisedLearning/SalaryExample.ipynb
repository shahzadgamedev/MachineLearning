# Regression Example
## Most Basic Example of Linear Regression
The purpose of this example is to understand the basic concept of linear regression. In this example, we will use the most basic form of linear regression with one feature (Years of Experience) and one target variable (Salary). We will calculate the cost function for different weights and biases and plot the cost function in 3D. We will also use gradient descent to find the optimal weight and bias values.

### Import Libraries
- Matplotlib for plotting graphs
- Matplotlib.cm for 3D plotting
- Numpy for numerical operations
- BasicFunctions for calculating cost function and model

```python
import matplotlib.pyplot as plt
from matplotlib import cm
import numpy as np
from Regression.BasicFunctions import BasicFunctions
```

### Import Data
- Import data from CSV file
- Ignore first row as it contains column names
- Split data into two columns (Years of Experience and Salary)

### Plot Data
- Plot data to visualize the relationship between Years of Experience and Salary
- Title: Salary vs Experience
- X-axis label: Years of Experience
- Y-axis label: Salary 

```python
data = np.genfromtxt(r'.\Regression\Data\Salary_Data.csv', delimiter=',', skip_header=1)
x = data[:, 0]
y = data[:, 1]

plt.scatter(x, y)
plt.title('Salary vs Experience')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

### Cost Function
- Calculate the cost function for given weight [8k to 10k] and bias [30k to 32k] for 2k samples
- Cost function is the mean squared error (Therefore single global minimum) [Convex Function]
- Mean squared error is the sum of squared differences between actual and predicted values
- Cost = 1/n * sum((y - y_pred)^2) [Basic Function Used]
- y_pred = m * x + b [Model in Basic Function Used]

```python
samples = 500
weights = np.linspace(8000, 10000, samples)
biases = np.linspace(28000, 31400, samples)
cost = np.zeros((samples, samples))

# Calculate cost for each weight and bias
for i, weight in enumerate(weights):
    for j, bias in enumerate(biases):
        y_pred = BasicFunctions.model(x, weight, bias)
        cost[i, j] = BasicFunctions.squared_error_cost(y, y_pred)

plt.subplots(1, 2, figsize=(16, 6))
# plot weight vs cost (weight on x-axis and cost on y-axis)
plt.subplot(1, 2, 1)
plt.plot(weights, cost[:, 0])
plt.xlabel('Weight')
plt.ylabel('Cost')
plt.title('Weight vs Cost')

# plot bias vs cost (bias on x-axis and cost on y-axis)
plt.subplot(1, 2, 2)
plt.plot(biases, cost[:, 1])
plt.xlabel('bias')
plt.ylabel('Cost')
plt.title('Bias vs Cost')

plt.tight_layout()
plt.show()

#plot cost function for different weight and bias
#plot 3D cost function
fig = plt.figure(figsize=(16, 10))
ax = fig.add_subplot(111, projection='3d')
X, Y = np.meshgrid(weights, biases)
ax.plot_surface(X, Y, cost, vmin=cost.min() * 0.5, cmap=cm.hsv, edgecolor='darkred', linewidth=0.1)
ax.set_xlabel('Weight')
ax.set_ylabel("Bias")
ax.set_zlabel("Cost")
# rotate the plot on y-axis
ax.view_init(azim=-160, elev=20)
plt.show()

```

### Gradient Descent
- using batch gradient descent (each iteration uses all samples)
- Calculate the gradient of the cost function
- Start with some w,b
- keep changing w,b to reduce cost
- until we reach the minimum (may have >1 minima)

```python
weight = 0
bias = 0
iterations = 1000
learning_rate = 0.0001

cost_history = np.zeros(iterations)
w_history = np.zeros(iterations)
b_history = np.zeros(iterations)

def gradient_weight(data_x, data_y, w, b) -> float:
    m = len(data_x)
    accumulated_error = 0.0
    for p in range(m):
        accumulated_error += (BasicFunctions.model(data_x[p], w, b) - data_y[p]) * data_x[p]
    return accumulated_error / m

def gradient_bias(data_x, data_y, w, b) -> float:
    m = len(data_x)
    accumulated_error = 0.0
    for p in range(m):
        accumulated_error += BasicFunctions.model(data_x[p], w, b) - data_y[p]
    return accumulated_error / m


# iterate to find optimal weight and bias
for i in range(iterations):
    temp_weight = weight - learning_rate * gradient_weight(x, y, weight, bias)
    temp_bias = bias - learning_rate * gradient_bias(x, y, weight, bias)
    weight, w_history[i] = temp_weight, temp_weight
    bias, b_history[i] = temp_bias, temp_bias
    cost_history[i] = BasicFunctions.squared_error_cost(y, BasicFunctions.model(x, weight, bias))

plt.subplots(1, 3, figsize=(25, 6))

plt.subplot(1, 3, 1)
plt.plot(range(iterations), cost_history)
plt.xlabel('Iterations')
plt.ylabel('Cost')
plt.title('Cost vs Iterations')


plt.subplot(1, 3, 2)
plt.plot(range(iterations), w_history)
plt.xlabel('Iterations')
plt.ylabel('Weight')
plt.title('Weight vs Iterations')

plt.subplot(1, 3, 3)
plt.plot(range(iterations), b_history)
plt.xlabel('Iterations')
plt.ylabel('Bias')
plt.title('Bias vs Iterations')


plt.tight_layout()
plt.show()
```
