# Regression Example
## Most Basic Example of Linear Regression

### Import Libraries
- Matplotlib for plotting graphs
- Numpy for numerical operations

```python
import matplotlib.pyplot as plt
import numpy as np
```

### Import Data
- Import data from CSV file
- Ignore first row as it contains column names
- Split data into two columns (Years of Experience and Salary)

```python
data = np.genfromtxt(r'F:\MachineLearning\MachineLearningPython\MachineLearning\SupervisedLearning\Regression\Data\Salary_Data.csv', delimiter=',', skip_header=1)
x = data[:, 0]
y = data[:, 1]
```

### Plot Data
- Plot data to visualize the relationship between Years of Experience and Salary
- Title: Salary vs Experience
- X-axis label: Years of Experience
- Y-axis label: Salary

```python
plt.scatter(x, y)
plt.title('Salary vs Experience')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

### Cost Function
- Calculate the cost function for given data points and parameters
- Cost function is the mean squared error
- Mean squared error is the sum of squared differences between actual and predicted values
- Cost = 1/n * sum((y - y_pred)^2)
- y_pred = m * x + b

```python
import SupervisedLearning.Regression.BasicFunctions as basicFunctions

weight = 9000
bias = 30000
n = len(x)
y_pred = np.zeros(n)
for i in range(n):
    y_pred[i] = basicFunctions.model(x[i], weight, bias)
    
cost = basicFunctions.squared_error_cost(y, y_pred)

#plotting the line of prediction vs actual data
plt.scatter(x, y)
plt.plot(x, y_pred, color='red')
plt.title('Salary vs Experience')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
print('Cost:', cost)

```
